```python
X = df.drop("Strength", axis=1)
y = df['Strength']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=15)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
#first weak learner
tree_reg1 = DecisionTreeRegressor(max_depth=3)
tree_reg1.fit(X_train,y_train)
```

```python
y2 = y_train - tree_reg1.predict(X_train)
```

```python
y2[:5] #residuals
```

```python
#second weak learner
tree_reg2 = DecisionTreeRegressor(max_depth=4)
tree_reg2.fit(X_train,y2)
```

```python
y3 = y2 -tree_reg2.predict(X_train)
y3[:5]
```

```python
#third weak learner
tree_reg3 = DecisionTreeRegressor(max_depth=4)
tree_reg3.fit(X_train,y3)
```

```python
y4 = y3 - tree_reg3.predict(X_train)
y4[:5]
```

```python
y_pred = sum(tree.predict(X_test) for tree in (tree_reg1,tree_reg2,tree_reg3))
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```

```python
#fourth weak learner
tree_reg4 = DecisionTreeRegressor(max_depth = 4)
tree_reg4.fit(X_train, y4)
y_pred = sum(tree.predict(X_test) for tree in (tree_reg1, tree_reg2, tree_reg3, tree_reg4))
```

```python
r2_score(y_test,y_pred)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
gbr = GradientBoostingRegressor(n_estimators=100,max_depth=3,learning_rate=0.1)
```

```python
gbr.fit(X_train,y_train)
y_pred = gbr.predict(X_test)
print(r2_score(y_test,y_pred))
```

```python
#hyperparameter tuning 
```

```python
params = {
    "n_estimators" : [100, 150, 200],
    "max_depth" : [3, 4, 5],
    "loss" : ["squared_error", "absolute_error", "huber", "quantile"],
    "learning_rate" : [0.01, 0.1, 0.5]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(estimator=GradientBoostingRegressor(),cv=5,param_grid=params,n_jobs=-1,verbose=3,scoring="r2")
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test)
print(r2_score(y_test,y_pred))
```
