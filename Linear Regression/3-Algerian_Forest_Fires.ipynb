```python
def correlation_for_dropping(df,threshold):
    columns_to_drop = set()
    corr = df.corr()
    for i in range(0,len(corr.columns)):
        for j in range(i):
            if abs(corr.iloc[i,j]) > threshold:
                columns_to_drop.add(corr.columns[i])
    return columns_to_drop
```

```python
correlation_for_dropping(X_train,0.15)
```

```python
correlation_for_dropping(X_train,0.95)
```

```python
columns_dropping = correlation_for_dropping(X_train,0.85)
```

```python
columns_dropping
```

```python
X_train.drop(columns_dropping, axis=1, inplace=True)
X_test.drop(columns_dropping, axis=1, inplace=True)
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
plt.subplots(figsize=(18,6))
plt.subplot(1,2,1)
sns.boxplot(data=X_train)
plt.title("X Train")
plt.subplot(1,2,2)
sns.boxplot(data=X_test_scaled)
plt.title("X Train Scaled")
plt.show()
```

```python
lin_regression = LinearRegression()
```

```python
lin_regression.fit(X_train_scaled,y_train)
```

```python
y_pred = lin_regression.predict(X_test_scaled)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test,y_pred)
print(f"MAE: {mae}, MSE:{mse}, R2 score: {score}")
plt.scatter(y_test,y_pred)
plt.show()
```

```python
lasso = Lasso() # default olarak cv = 1
lasso.fit(X_train_scaled,y_train)
y_pred = lasso.predict(X_test_scaled)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test,y_pred)
print(f"MAE: {mae}, MSE:{mse}, R2 score: {score}")
plt.scatter(y_test,y_pred)
plt.show()
```

```python
ridge = Ridge()
ridge.fit(X_train_scaled,y_train)
y_pred = ridge.predict(X_test_scaled)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test,y_pred)
print(f"MAE: {mae}, MSE:{mse}, R2 score: {score}")
plt.scatter(y_test,y_pred)
plt.show()
```

```python
elastic_net = ElasticNet()
elastic_net.fit(X_train_scaled,y_train)
y_pred = elastic_net.predict(X_test_scaled)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test,y_pred)
print(f"MAE: {mae}, MSE:{mse}, R2 score: {score}")
plt.scatter(y_test,y_pred)
plt.show()
```
