```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("1-bank_customers.csv")
```

```python
df.head() #scaled and encoded
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
X = df.drop("subscribed",axis=1)
y = df["subscribed"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=15)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic = LogisticRegression()
logistic.fit(X_train,y_train)
y_pred = logistic.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report #recall_score, precision_score
```

```python
score = accuracy_score(y_pred,y_test)
print("score: ",score)
print(classification_report(y_pred,y_test))
print("confusion matrix: \n",confusion_matrix(y_pred,y_test))
```

```python
#hyperparameter tuning
```

```python
model = LogisticRegression() 
```

```python
#  L1 (Lasso Regression) → feature selection 
#  L2 (Ridge Regression) → multicollinearity .
```

```python
penalty = ["l1","l2","elasticnet"]
c_values = [100, 10, 1, 0.1, 0.01]
solver = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
parameters = dict(penalty = penalty, C = c_values, solver = solver)
```

```python
parameters
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
```

```python
cv = StratifiedKFold(n_splits = 5)
```

```python
grid = GridSearchCV(estimator=model, param_grid = parameters, cv=cv, scoring = "accuracy",n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_ 
```

```python
grid.best_score_ 
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print("score: ",score)
print(classification_report(y_pred,y_test))
print("confusion matrix: \n",confusion_matrix(y_pred,y_test))
```

```python
from sklearn.model_selection import RandomizedSearchCV
```
