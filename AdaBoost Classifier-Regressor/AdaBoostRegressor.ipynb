```python
X = df.drop("selling_price",axis=1)
y = df['selling_price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=15)
```

```python
categorical_cols = [col for col in df.columns if df[col].dtype == "object"]
# categorical_cols = df.select_dtypes("object").columns.to_list()
```

```python
unique_values = df[categorical_cols].nunique()
print(unique_values)
```

```python
# seller_type, fuel_type, transmission_type -> One Hot encoding
# car_name, brand, model -> frequency encoding 
```

```python
onehot_columns = ['seller_type', 'fuel_type', 'transmission_type']
freq_columns = ['car_name', 'brand', 'model']
```

```python
for col in freq_columns:
    # calculate frequencies in training
    freq = X_train[col].value_counts() / len(X_train) # frequency percentage

    X_train[col + '_freq'] = X_train[col].map(freq)

    X_test[col + '_freq'] = X_test[col].map(freq)
    
    # if there is a different category that is left nan in X_test just fill it with mean
    mean_freq = freq.mean()
    X_test[col + '_freq'] = X_test[col + '_freq'].fillna(mean_freq)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
X_train = X_train.drop(columns=["car_name","brand","model"],axis=1)
X_test = X_test.drop(columns=["car_name","brand","model"],axis=1)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(
    transformers=[
        ("OneHot", OneHotEncoder(drop="first", handle_unknown="ignore"), onehot_columns)
    ],
    remainder="passthrough"
)

X_train = transformer.fit_transform(X_train)
X_test = transformer.transform(X_test)

encoded_columns = transformer.get_feature_names_out()

X_train = pd.DataFrame(data=X_train, columns=encoded_columns)
X_test = pd.DataFrame(data=X_test, columns=encoded_columns)
```

```python
X_train.head()
```

```python
X_train.info()
```

```python
from sklearn.ensemble import AdaBoostRegressor, RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVR
from sklearn.neighbors import  KNeighborsRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.model_selection import RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
```

```python
def calculate_model_metric(true, predicted):
    mae = mean_absolute_error(true, predicted)
    mse = mean_squared_error(true, predicted)
    rmse = np.sqrt(mean_squared_error(true, predicted))
    score = r2_score(true, predicted)
    return mae, rmse, score
```

```python
model = AdaBoostRegressor()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)
model_train_mae, model_train_rmse, model_train_r2 = calculate_model_metric(y_train,y_train_pred)
model_test_mae, model_test_rmse, model_test_r2 = calculate_model_metric(y_test,y_test_pred)

print("Evaluation for Training Set")
print("RMSE :", model_train_rmse)
print("Mean Absolute Error :", model_train_mae)
print("R2 Score :", model_train_r2)
print("-----------------------------")

print("Evaluation for Test Set")
print("RMSE :", model_test_rmse)
print("Mean Absolute Error :", model_test_mae)
print("R2 Score :", model_test_r2)
```

```python
params = {
    "n_estimators" : [50, 80, 100, 120],
    "learning_rate" : [0.001, 0.01, 0.1, 1.0, 2.0],
    "loss" : ["linear", "square", "exponential"]
}
```

```python
rcv = RandomizedSearchCV(estimator=AdaBoostRegressor(),param_distributions=params,cv=5,n_iter=20,n_jobs=-1,scoring="r2",verbose=3)
```

```python
rcv.fit(X_train,y_train)
```

```python
rcv.best_params_
```

```python
y_pred = rcv.predict(X_test)

y_train_pred = rcv.predict(X_train)
y_test_pred = rcv.predict(X_test)

model_train_mae, model_train_rmse, model_train_r2 = calculate_model_metric(y_train,y_train_pred)
model_test_mae, model_test_rmse, model_test_r2 = calculate_model_metric(y_test,y_test_pred)

print("Evaluation for Training Set")
print("RMSE :", model_train_rmse)
print("Mean Absolute Error :", model_train_mae)
print("R2 Score :", model_train_r2)
print("-----------------------------")

print("Evaluation for Test Set")
print("RMSE :", model_test_rmse)
print("Mean Absolute Error :", model_test_mae)
print("R2 Score :", model_test_r2)
```
