```python
#top app categories by installment
df_grouped_by_category = df_clean.groupby(["Category"])
df_grouped_by_category["Installs"].sum().sort_values(ascending=False)
```

```python
new_df = df_grouped_by_category["Installs"].sum().sort_values(ascending=False).reset_index()
new_df["Installs"] = new_df["Installs"] /1_000_000_000 
print(new_df)
```

```python
#TOP 10 CATEGORİES BY INSTALLS
df2 = new_df.head(10)
plt.figure(figsize=(10,5))

sns.barplot(x="Installs",y="Category",data=df2)
plt.tight_layout()
plt.show()
```

```python
#Top 5 app
df_clean.groupby(["Category","App"])["Installs"].sum()
```

```python
df_groubed_by_app = df_clean.groupby(["Category","App"])["Installs"].sum().reset_index()
df_groubed_by_app.sort_values("Installs",ascending=False,inplace=True)
print(df_groubed_by_app)
```

```python
apps = ["GAME","COMMUNICATION","TOOLS","PRODUCTIVITY","SOCIAL"]
plt.figure(figsize=(40,40))
for i, app in enumerate(apps):
    df3 = df_groubed_by_app[df_groubed_by_app.Category == app]
    df3 = df3.head(5)
    print(df3)
    plt.subplot(3,2,i+1)
    sns.barplot(data=df3,x="Installs",y="App")
    plt.title(app,size = 15)
plt.tight_layout() 
plt.show()
```
